### Install the Required Libraries


### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset into a DataFrame



load the dataset using pd.read_csv()


```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
# 1. Contar el número de filas duplicadas en el dataset
num_duplicates = df.duplicated().sum()
print(f"Número total de filas duplicadas: {num_duplicates}")

# 2. Mostrar las primeras filas duplicadas
duplicate_rows = df[df.duplicated()]
print("\nEstructura de las primeras filas duplicadas:")
duplicate_rows.head()
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
# 1. Eliminar las filas duplicadas del dataset
df.drop_duplicates(inplace=True)

# 2. Verificar la eliminación contando de nuevo los duplicados
num_duplicates_after = df.duplicated().sum()
print(
    f"Número de filas duplicadas después de la eliminación: {num_duplicates_after}"
)
print(f"Nuevo tamaño del DataFrame: {df.shape}")
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
# 1. Identificar valores faltantes por columna
print("Valores faltantes por columna:")
print(df.isnull().sum())

# 2. Encontrar el valor más frecuente (moda) en 'EdLevel' e imputarlo
most_frequent_edlevel = df['EdLevel'].mode()[0]
df['EdLevel'] = df['EdLevel'].fillna(most_frequent_edlevel)

print(f"\nValor más frecuente en 'EdLevel': {most_frequent_edlevel}")
print(f"Valores faltantes restantes en 'EdLevel': {df['EdLevel'].isnull().sum()}")
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
# 1. Verificar valores faltantes en la columna ConvertedCompYearly
missing_comp = df["ConvertedCompYearly"].isnull().sum()
print(f"Valores faltantes en 'ConvertedCompYearly': {missing_comp}")

# 2. Manejar valores faltantes imputando la mediana sin usar inplace=True
median_comp = df["ConvertedCompYearly"].median()
df["ConvertedCompYearly"] = df["ConvertedCompYearly"].fillna(median_comp)

# Verificar que ya no queden valores nulos
print(
    f"Valores nulos restantes en 'ConvertedCompYearly': {df['ConvertedCompYearly'].isnull().sum()}"
)
```
