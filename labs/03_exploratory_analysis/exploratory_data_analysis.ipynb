#### Step 1: Install and Import Libraries


Install the necessary libraries for data manipulation and visualization.


```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
# Rellenar 'Employment' y 'RemoteWork' con la moda (valor más frecuente)
df['Employment'] = df['Employment'].fillna(df['Employment'].mode()[0])
df['RemoteWork'] = df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])

# Rellenar 'JobSat' con la mediana o moda según aplique
df['JobSat'] = df['JobSat'].fillna(df['JobSat'].median())

# Verificar que no queden nulos en esas columnas
print(df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Convertir YearsCodePro a tipo numérico
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# Convertir JobSat a tipo numérico por si está en formato de texto
df['JobSat'] = pd.to_numeric(df['JobSat'], errors='coerce')

# Crear rangos/categorías de experiencia
bins = [-1, 5, 10, 20, 100]
labels = ['0-5 years', '5-10 years', '10-20 years', '>20 years']
df['ExperienceRange'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels)

# Calcular la mediana de JobSat por rango de experiencia
median_jobsat = df.groupby('ExperienceRange')['JobSat'].median().reset_index()
print(median_jobsat)

# Visualizar la relación mediante un gráfico de barras
plt.figure(figsize=(8, 5))
sns.barplot(data=median_jobsat, x='ExperienceRange', y='JobSat')
plt.title('Median Job Satisfaction by Experience Range')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Job Satisfaction')
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
# Gráfico de conteo (count plot) para la distribución de valores en JobSat
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='JobSat')
plt.title('Distribution of Job Satisfaction Scores')
plt.xlabel('Job Satisfaction')
plt.ylabel('Count')
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Gráfico de conteo para la distribución del trabajo remoto
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='RemoteWork')
plt.title('Distribución de Trabajo Remoto')
plt.xlabel('Remotework')
plt.ylabel('Cantidad')
plt.show()

# 2. Tabulación cruzada entre preferencia de trabajo remoto y tipo de empleo
ct_remote = pd.crosstab(df['Employment'], df['RemoteWork'])
print("Tabulación cruzada (Employment vs RemoteWork):")
print(ct_remote)
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Separar los lenguajes y expandir las filas
df_lang = df[['Country', 'LanguageHaveWorkedWith']].dropna().copy()
df_lang['LanguageHaveWorkedWith'] = df_lang['LanguageHaveWorkedWith'].str.split(';')

# Expande las listas y resetea el índice para evitar duplicados en el axis
df_exploded = df_lang.explode('LanguageHaveWorkedWith').reset_index(drop=True)

# 2. Obtener los 5 países principales y los 10 lenguajes más populares
top_countries = df_exploded['Country'].value_counts().head(5).index
top_languages = df_exploded['LanguageHaveWorkedWith'].value_counts().head(10).index

# Filtrar el DataFrame
df_filtered = df_exploded[
    df_exploded['Country'].isin(top_countries) & 
    df_exploded['LanguageHaveWorkedWith'].isin(top_languages)
]

# 3. Crear la tabulación cruzada (crosstab) y la gráfica
ct_lang = pd.crosstab(df_filtered['Country'], df_filtered['LanguageHaveWorkedWith'])

plt.figure(figsize=(10, 6))
sns.heatmap(ct_lang, annot=True, fmt='d', cmap='Blues')
plt.title('Popularidad de Lenguajes de Programación por Región/País')
plt.xlabel('Lenguaje de Programación')
plt.ylabel('País')
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
# Convertir columnas a tipo numérico por seguridad
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df['JobSatPoints_1'] = pd.to_numeric(df['JobSatPoints_1'], errors='coerce')

# Gráfico de dispersión (Scatter Plot)
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='YearsCodePro', y='JobSatPoints_1', alpha=0.5)
plt.title('Correlación entre Años de Experiencia y Satisfacción Laboral')
plt.xlabel('Años de Experiencia (YearsCodePro)')
plt.ylabel('Puntos de Satisfacción (JobSatPoints_1)')
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Tabulación cruzada
ct_edu = pd.crosstab(df['EdLevel'], df['Employment'])
print("Tabulación cruzada (EdLevel vs Employment):")
print(ct_edu)

# 2. Visualización
fig, ax = plt.subplots(figsize=(12, 6))
sns.countplot(data=df, y='EdLevel', hue='Employment', ax=ax)

plt.title('Relación entre Nivel Educativo y Tipo de Empleo')
plt.xlabel('Cantidad')
plt.ylabel('Nivel Educativo')

# Mover la leyenda fuera sin apretar el gráfico
plt.legend(title='Employment', bbox_to_anchor=(1.02, 1), loc='upper left')

# Mostrar ajustando los bordes adecuadamente
plt.savefig('temp_plot.png', bbox_inches='tight')
plt.show()
```
